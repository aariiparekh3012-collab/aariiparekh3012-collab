```python
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns

corr = df[numeric_cols].corr()

corr.round(2)
```

```python
categorical_cols = [
    "gender",
    "occupation_type",
    "chronotype",
    "primary_bedtime_app"
]

for col in categorical_cols:
    print(f"\n=== {col} ===")
    display(
        pd.crosstab(
            df[col],
            df["sleep_debt_category"],
            normalize="index"
        ).round(3)
    )
```
